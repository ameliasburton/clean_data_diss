```python
"""
Training loop for the EGFR GNN regressor.
"""
import torch
from torch_geometric.data import DataLoader
from model_arch import EGFR_GNN_Regressor
from data_prep import load_dataset
import torch.optim as optim
import torch.nn as nn

def train_epoch(model, loader, optimizer, device='cpu'):
    model.train()
    losses = []
    for batch in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        out = model(batch.x, batch.edge_index, batch.batch, edge_attr=batch.edge_attr, global_features=batch.global_features)
        loss = nn.MSELoss()(out, batch.y)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())
    return sum(losses) / len(losses) if losses else 0.0

if __name__ == "__main__":
    dataset = load_dataset('data/processed/egfr_train.pkl')
    loader = DataLoader(dataset, batch_size=32, shuffle=True)
    model = EGFR_GNN_Regressor()
    optimizer = optim.Adam(model.parameters(), lr=1e-3)
    for epoch in range(10):
        loss = train_epoch(model, loader, optimizer)
        print(f'Epoch {epoch}: loss={loss:.4f}')
```
